```
# Understand the sentiment of customer review using Cortex Sentiment
reviews_df = session.table('DAILY_REVENUE')

reviews_df
```

```
product_dim = session.table("PRODUCT_DIM")

product_dim
```

```
region_dim = session.table("REGION_DIM")

region_dim
```

```
from snowflake.snowpark.functions import col

# Join entre reviews_df y product_dim usando PRODUCT_ID
joined_df = (
    reviews_df.join(product_dim, reviews_df["PRODUCT_ID"] == product_dim["PRODUCT_ID"], "left")
              .join(region_dim, region_dim["REGION_ID"] == region_dim["REGION_ID"], "left")
)

# Mostrar resultados
joined_df
```
